```python
from feynpy import *
print("FeynPy imports OK ✅")
```

```python
# Define symbols
θ, φ = sp.symbols('theta phi', real=True)
p = sp.symbols('p', real=True, positive=True)
```

```python
p1 = MomUP(p, p, 0, 0)          # e-
p2 = MomUP(p, p, sp.pi, 0)      # e+
p3 = MomUP(p, p, θ, 0)          # mu-
p4 = MomUP(p, p, sp.pi+ θ, 0)   # mu+
display(p1, p2, p3, p4)
```

```python
def Amplitude(h1, h2, h3, h4):
    e = sp.symbols('e', real=True)

    s = minkowski_dot(p1 + p2, p1 + p2)

    u1 = u_spinor(h1, p, p, 0, 0)
    v2 = v_spinor(h2, p, p, sp.pi, 0)
    u3 = u_spinor(h3, p, p, θ, 0)
    v4 = v_spinor(h4, p, p, sp.pi+ θ, 0)

    M = sp.ZeroMatrix(1, 1)
    for mu in range(4): 
        M+= - e**2 / s * \
        (Bar(v2) * GammaUP(mu) * Gamma6 * u1 + Bar(v2) * GammaUP(mu) * Gamma7 * u1) * \
        (Bar(v4) * GammaDOWN(mu) * Gamma6 * u3 + Bar(v4) * GammaDOWN(mu) * Gamma7 * u3)
        

    return sp.simplify(M)

```

```python
h1 = +1/2  # e- helicity
h2 = -1/2  # e+ helicity
h3 = -1/2  # mu- helicity
h4 = +1/2  # mu+ helicity

M = Amplitude(h1, h2, h3, h4)

def hel_to_pm(h):
    return '+' if h == +1/2 else '-'

for h1 in [+1/2, -1/2]:
    for h2 in [+1/2, -1/2]:
        for h3 in [+1/2, -1/2]:
            for h4 in [+1/2, -1/2]:
                M = Amplitude(h1, h2, h3, h4)

                print(f"Helicities: {hel_to_pm(h1)}, {hel_to_pm(h2)}, {hel_to_pm(h3)}, {hel_to_pm(h4)}")
                display(M)
                print("--------------------------------")
                print("\n")
```
