```python
from feynpy import *
print("FeynPy imports OK ✅")
```

```python
# Define symbols
θ, φ = sp.symbols('theta phi', real=True)
s, mz = sp.symbols('s mz', real=True, positive=True)
```

```python
p1 = MomUP(sp.sqrt(s)/2, sp.sqrt(s)/2, 0, 0)          # e-
p2 = MomUP(sp.sqrt(s)/2, sp.sqrt(s)/2, sp.pi, 0)      # e+
p3 = MomUP(sp.sqrt(s)/2, sp.sqrt(s/4 - mz**2), θ, 0)          # Z
p4 = MomUP(sp.sqrt(s)/2, sp.sqrt(s/4 - mz**2), sp.pi+ θ, 0)   # Z
display(p1, p2, p3, p4)
```

```python
def Amplitude(h1, h2, λ3, λ4):
    e, α, theta_W, MZ2, s = sp.symbols('e α theta_W MZ2 s', real=True)
    u1 = u_spinor(h1, sp.sqrt(s)/2, sp.sqrt(s)/2, 0, 0)
    v2 = v_spinor(h2, sp.sqrt(s)/2, sp.sqrt(s)/2, sp.pi, 0)
    eps3 = EpsilonUP(λ3, sp.sqrt(s)/2, sp.sqrt(s/4 - mz**2), θ, 0)
    eps4 = EpsilonUP(λ4, sp.sqrt(s)/2, sp.sqrt(s/4 - mz**2), sp.pi+ θ, 0)
    
    t = minkowski_dot(p1 - p3, p1 - p3)
    u = minkowski_dot(p1 - p4, p1 - p4)

    
    M = - 1 / (4 * (sp.cos(theta_W)**2) * (sp.sin(theta_W)**2)) * (
            4 * sp.sin(theta_W)**4  * \
                (
                    -2 * (
                        minkowski_dot(p2, eps3) * Bar(v2) * Slash(eps4) * Gamma6 * u1 / u + \
                        minkowski_dot(p2, eps4) * Bar(v2) * Slash(eps3) * Gamma6 * u1 / t \
                    ) + \
                    Bar(v2) * Slash(eps3) * Slash(p3) * Slash(eps4) * Gamma6 * u1 / u + \
                    Bar(v2) * Slash(eps4) * Slash(p4) * Slash(eps3) * Gamma6 * u1 / t \
                ) - \
            2 * (1 - 2 * sp.sin(theta_W)**2)**2 * minkowski_dot(p2, eps3) * Bar(v2) * Slash(eps4) * Gamma7 * u1 / u - \
            2 * (1 - 2 * sp.sin(theta_W)**2)**2 * minkowski_dot(p2, eps4) * Bar(v2) * Slash(eps3) * Gamma7 * u1 / t + \
            (1 - 2 * sp.sin(theta_W)**2)**2 * Bar(v2) * Slash(eps3) * Slash(p3) * Slash(eps4) * Gamma7 * u1 / u + \
            (1 - 2 * sp.sin(theta_W)**2)**2 * Bar(v2) * Slash(eps4) * Slash(p4) * Slash(eps3) * Gamma7 * u1 / t 
        )
    
    
    return sp.simplify(M)
```

```python
h1 = +1/2  # e- helicity
h2 = -1/2   # e+ helicity
λ1 = -1  # Z helicity
λ2 = +1  # Z helicity

M = Amplitude(h1, h2, λ1, λ2)

def hel_to_pm(h):
    return '+' if h == +1/2 or h == +1 else '-' if h == -1/2 or h == -1 else '0'


for h1 in [+1/2, -1/2]:
    for h2 in [+1/2, -1/2]:
        for λ1 in [+1,-1]:
            for λ2 in [+1,-1]:
                M = Amplitude(h1, h2, λ1, λ2)

                print(f"Helicities: {hel_to_pm(h1)}, {hel_to_pm(h2)}, {hel_to_pm(λ1)}, {hel_to_pm(λ2)}")
                display(M), display(sp.simplify(M.subs(mz,0)))
                print("--------------------------------")
                print("\n")
```
