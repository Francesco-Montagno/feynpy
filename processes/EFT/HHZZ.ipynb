```python
# Define symbols
θ, φ = sp.symbols('theta phi', real=True)
p, En, q = sp.symbols('p En q', real=True, positive=True)
```

```python
p1 = MomUP(p, p, 0, 0)          # e-
p2 = MomUP(p, p, sp.pi, 0)      # e+
p3 = MomUP(En, q, θ, 0)          # Z
p4 = MomUP(En, q, sp.pi+ θ, 0)   # Z
display(p1, p2, p3, p4)
```

```python
def Amplitude(h1, h2, λ3, λ4):
    u1 = u_spinor(h1, p, p, 0, 0)
    v2 = v_spinor(h2, p, p, sp.pi, 0)
    eps1 = EpsilonUP(λ3, En, q, θ, 0)
    eps2 = EpsilonUP(λ4, En, q, sp.pi+ θ, 0)

    y, mH, G = sp.symbols('y mH G', real=True)

    s = minkowski_dot(p1 + p2, p1 + p2)
    t = minkowski_dot(p1 - p3, p1 - p3)
    u = minkowski_dot(p1 - p4, p1 - p4)
    M = sp.ZeroMatrix(1, 1)
    M = Bar(v2) * (-I * y) * u1 * \
        I / (s - mH**2) * \
        G *(    minkowski_dot(p3,eps1.conjugate()) * minkowski_dot(p4,eps2.conjugate()) +  \
                minkowski_dot(p4,eps1.conjugate()) * minkowski_dot(p3,eps2.conjugate()) - \
            2 * minkowski_dot(eps1.conjugate(),eps2.conjugate()) * (minkowski_dot(p3,p4))
            )

    return sp.simplify(M)

```

```python
h1 = +1/2  # e- helicity
h2 = -1/2   # e+ helicity
λ1 = -1  # Z helicity
λ2 = +1  # Z helicity

M = Amplitude(h1, h2, λ1, λ2)

def hel_to_pm(h):
    return '+' if h == +1/2 or h == +1 else '-' if h == -1/2 or h == -1 else '0'


for h1 in [+1/2, -1/2]:
    for h2 in [+1/2, -1/2]:
        for λ1 in [+1, 0, -1]:
            for λ2 in [+1, 0, -1]:
                M = Amplitude(h1, h2, λ1, λ2)

                print(f"Helicities: {hel_to_pm(h1)}, {hel_to_pm(h2)}, {hel_to_pm(λ1)}, {hel_to_pm(λ2)}")
                display(M.simplify())
                print("--------------------------------")
                print("\n")
```
